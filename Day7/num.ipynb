***matplotlib***

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x=np.random.randint(1,50,10)
y=np.random.randint(1,50,10)
print(x,y,sep='\n')
```

```python
plt.plot(x,y)
```

```python
plt.plot(x,y,color='red',linestyle='dotted')
plt.show()
```

```python
plt.title("Random chart")
plt.plot(x,y,color='red',linestyle='dotted')
plt.show()
```

```python
plt.title("Random chart")
plt.xlabel("x-axis label")
plt.ylabel("y-axis label")
plt.plot(x,y,color='red',linestyle='dotted')
plt.show()
```

```python
plt.title("Random chart")
plt.xlabel("x-axis label")
plt.ylabel("y-axis label")
plt.xticks(x)
plt.yticks(y)
plt.plot(x,y,color='red',linestyle='dotted')
plt.show()
```

- -> solid(default) 
-- -> dashed 
-. -> dash-dot 
: -> dotted

```python
plt.title("Random chart")
plt.xlabel("x-axis label")
plt.ylabel("y-axis label")
plt.xticks(x)
plt.yticks(y)
plt.plot(x,y,color='red',linestyle='-',alpha=1)
plt.show()
```

```python
plt.title("Random chart")
plt.xlabel("x-axis label")
plt.ylabel("y-axis label")
plt.xticks(x)
plt.yticks(y)
plt.plot(x,y,color='red',linestyle='-',alpha=0.9,
         marker='o')
plt.show()
```

```python
plt.title("Random chart")
plt.xlabel("x-axis label")
plt.ylabel("y-axis label")
plt.xticks(x)
plt.yticks(y)
plt.plot(x,y,color='red',linestyle='-',alpha=0.9,
         marker='o',markerfacecolor='b')
plt.show()
```

```python
plt.title("Random chart")
plt.xlabel("x-axis label")
plt.ylabel("y-axis label")
plt.xticks(x)
plt.yticks(y)
plt.plot(x,y,color='red',linestyle='-',alpha=0.9,
         marker='o',markerfacecolor='b',markersize=10)
plt.show()
```

```python
plt.title("Random chart")
plt.xlabel("x-axis label")
plt.ylabel("y-axis label")
plt.xticks(x)
plt.yticks(y)
plt.plot(x,y,color='red',linestyle='-',alpha=0.9,
         marker='o',markerfacecolor='b',
         markersize=10,markeredgecolor='black')
plt.show()
```

```python
plt.title("Random chart")
plt.xlabel("x-axis label")
plt.ylabel("y-axis label")
plt.xticks(x)
plt.yticks(y)
plt.plot(x,y,color='red',linestyle='-',alpha=0.9,
         marker='o',markerfacecolor='b',
         markersize=10,markeredgecolor='black',
         linewidth=3)
plt.show()
```

```python
x=np.arange(1,11)
y=np.sin(x)
z=np.cos(x)
print(x,y,z,sep='\n\n')
```

```python
plt.figure(figsize=(8,4))
plt.plot(x,y,label='SINQ',color='red')
plt.plot(x,z,label='COSQ',color='black')
plt.title("SINQ & COSQ")
plt.xlabel=('Data points')
plt.ylabel=("Associate value")
plt.xticks()
plt.yticks()
plt.legend()
plt.show()
```

```python
plt.plot(x,z)
```

```python
plt.figure(figsize = (8, 4))

plt.subplot(1, 2, 1)
plt.plot(x, y, label='SINQ', color='red')
plt.plot(x, z, label='COSQ')
plt.title("SINQ & COSQ")
plt.xlabel("Data Points")
plt.ylabel("Associate Values")
plt.legend()
plt.grid()

plt.subplot(1, 2, 2)
plt.plot(x, y, label='SINQ', color='green')
plt.plot(x, z, label='COSQ', color='red')
plt.title("SINQ & COSQ")
plt.xlabel("Data Points2")
plt.ylabel("Associate Values2")
plt.subplots_adjust=0.8
plt.legend()
plt.grid()

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize = (8, 4))

plt.subplot(3, 3, 1)
plt.plot(x, y, label='SINQ', color='red')
plt.plot(x, z, label='COSQ')
plt.title("SINQ & COSQ")
plt.xlabel("Data Points")
plt.ylabel("Associate Values")
plt.legend()

plt.subplot(3, 3, 5)
plt.plot(x, y, label='SINQ', color='green')
plt.plot(x, z, label='COSQ', color='red')
plt.title("SINQ & COSQ")
plt.xlabel("Data Points2")
plt.ylabel("Associate Values2")
plt.legend()

plt.subplot(3, 3, 9)
plt.plot(x, y, label='SINQ', color='green')
plt.plot(x, z, label='COSQ', color='red')
plt.title("SINQ & COSQ")
plt.xlabel("Data Points3")
plt.ylabel("Associate Values3")
plt.legend()

plt.tight_layout()
plt.show()
```
