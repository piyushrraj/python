```python
#for count the number of words in file
with open('data.txt') as f:
    words=f.read().split()
    print(len(words))
```

```python
#unique word put value in set
with open('data.txt') as f:
    set1=set(f.read().split())
    print(set1)
    print(len(set1))
```

```python
dic={}
with open ('data.txt') as f:
    lis=f.read().split()
    for i in lis:
        if i in dic:
            dic[i]+=1
        else:
            dic[i]=1
print(dic)
```

***json file***

```python
data={
    "name":"Aditya",
    "age": 22,
    "course":"CSE"
}
```

```python
import json
with open('json1.json','w') as f:
    json.dump(data,f)
    
```

```python
with open('json1.json','r') as f:
    
    print(json.load(f))
```

```python

with open('json1.json','w+') as f:
    json.dump(data,f)
    f.seek(0)
    print(json.load(f))
```

```python
import json
with open('json1.json', 'r+') as f:
    json.dump(data, f)
    f.seek(0)          
    da = json.load(f)
    print(da)
    print(len(da))
```

***CSV***

```python
import csv
with open("data.csv",'w',newline="") as file:
    writer=csv.writer(file)
    writer.writerow(["Name","Age","Course"])
    writer.writerow(["Aditya",23,"cse"])
```

```python
rows=[
    ["Name","Age"],
    ["Aman",23],
    ["abc",36]
]
with open ("data.csv",'w',newline="")as file:
    writer=csv.writer(file)
    writer.writerows(rows)
```

```python
#to create header and csv.dictwriter()
import csv
with open("data.csv","w",newline="")as file:
    fieldnames=["Name","age"]
    writer=csv.DictWriter(file,fieldnames=fieldnames)
    writer.writeheader()
    writer.writerow({"Name":"abc","age":21})
    writer.writerow({"Name":"abcd","age":23}) 
```

```python
with open("data.csv","r")as file:
    reader=csv.DictReader(file)
    print(reader)
    for row in reader:
        print(row)
```
