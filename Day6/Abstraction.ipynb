```python
from abc import ABC, abstractmethod
class BankAccount(ABC):
    def __init__(self,customer_name,date_of_birth):
        self.customer_name=customer_name
        self.date_of_birth=date_of_birth
    @abstractmethod
    def deposit(self,amount):
        pass
    @abstractmethod
    def withdraw(self,amount):
        pass
    
```

```python
class SavingAccount(BankAccount):
    def __init__(self, customer_name, date_of_birth,minimum_balance,balance):
        super().__init__(customer_name, date_of_birth)
        self.minimum_balance=minimum_balance
        self.balance=balance
        
    
    def deposit(self, amount):
        if(amount>0):
            self.balance+=amount
            print(f"{amount} deposit successfully")
            print(f"current amount {self.balance}")
        else:
            print("amount should be positive")
    def withdraw(self, amount):
        if(amount>self.balance-self.minimum_balance):
            print("amount should be less")
        else:
            self.balance-=amount
            print(f"remaining amount {self.balance}")
     
```

```python
obj1=SavingAccount("abc","12-02-11",1000,0)
obj1.withdraw(1000)
obj1.deposit(10000)
```

```python
class CurrentAccount(BankAccount):
    def __init__(self, customer_name, date_of_birth,overdraft,balance):
        super().__init__(customer_name, date_of_birth)
        self.overdraft=overdraft
        self.balance=balance
        
    
    def deposit(self, amount):
        if(amount>0):
            self.balance+=amount
            print(f"{amount} deposit successfully")
            print(f"current amount {self.balance}")
        else:
            print("amount should be positive")
    def withdraw(self, amount):
        if(amount<self.balance):
            self.balance-=amount
            print(f"remaining amount {self.balance}")
        elif(amount<self.balance+self.overdraft):
            self.overdraft-=amount-self.balance
            self.balance=0
            print(f"draft amount remaining {self.overdraft} and remaining main balance = {self.balance}")
        else:
            print("amount should be less")
     
```

```python
obj2=CurrentAccount("Piyush","12-04-2004",400,0)
obj2.deposit(1000)
obj2.withdraw(1200)
```
