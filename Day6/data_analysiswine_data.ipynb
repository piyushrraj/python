Merge and concat

```python
data1={
    'Name':['A','B','C'],
    'age':[23,25,67],
    'salary':[1000,2000,3000]
}
data2={
    'Name':['A','BB','C'],
    'age':[23,23,67],
    'salary':[1000,2300,3000]
}
```

```python
new_df1=pd.DataFrame(data1)
new_df2=pd.DataFrame(data2)
```

```python
new_df1.head()
```

```python
pd.concat([new_df1,new_df2])
```

```python
pd.concat([new_df1,new_df2],ignore_index=True)
```

```python
#column-wise concatinate by using axis=1 
new_def=pd.concat([new_df1,new_df2],ignore_index=True,axis=0)
new_def
```

```python
new_def.drop_duplicates(inplace=True)
new_def.head()
```
