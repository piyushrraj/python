```python
class ten:
    def __init__(self, name, age, marks):
        self.name=name
        self.age=age
        self.marks=marks
    def display(self):
        print(self.name,self.age,self.marks)

        
    
class twelve(ten):
    def __init__(self, name, age, marks,stream):
        super().__init__(name, age, marks)
        self.stream=stream
    def display(self):
        print(self.name,self.age,self.stream)

obj1=twelve("abc",17,378,"maths")
obj1.display()
        


```

```python
class bach(twelve):
    def __init__(self, name, age, marks, stream,branch):
        super().__init__(name, age, marks, stream)
        self.branch=branch
    def display(self):
        print(self.name,self.age,self.stream,self.branch)
obj2=bach("hello",18,637,"science","cse")
obj2.display()
```

```python
class person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def display(self):
        print(self.name,self.age)

class student(person):
    def __init__(self, name, age,school):
        super().__init__(name, age)
        self.school=school
    def display(self):
        print(self.name,self.age,self.school)
    
class teacher(person):
    def __init__(self, name, age,subject_teach):
        super().__init__(name, age)
        self.subject_teach=subject_teach
    def display(self):
        print(self.name,self.age,self.school,self.subject_teach)

class academic(student):
    def __init__(self,name,age,school,subject,marks):
        super().__init__(name,age,school)
        self.subject=subject
        self.marks=marks
    def display(self):
        print(self.name,self.age,self.school,self.subject,self.marks)

class sports(student):
    def __init__(self,sport_name,level,school,name,age):
        super().__init__(name,age,school)
        self.sport_name=sport_name
        self.level=level
    def display(self):
        print(self.name,self.age,self.school,self.sport_name,self.level)

class allRounderStudent(academic,sports):
    def __init__(self, name,age,school,marks,subject,sport_name,level):
        super().__init__(name,age,school,marks,subject, marks,sport_name,level)
    def display(self):
        print(self.name,self.age,self.school,self.marks,self.subject,self.sport_name,self.level)
a=allRounderStudent("abc",17,"abc_school","maths",98,"cricket",8)
a.display()
    



        
```
